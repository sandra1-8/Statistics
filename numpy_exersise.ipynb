```python
import numpy as np
arr1 = np.array([1,2,3,4,5]) #1D array vector
arr2 = np.array([[1,2,3],[4,5,6]]) #2D array i.e matrix

print(arr1)
print(arr2)
```

```python
# creating array with zero and ones

a = np.zeros(5) # 1D array of zeros
b = np.ones((3,3))  # 2D array of zeros

print(a)
print(b)
```

```python
a = np.array([1,2,3,4,5])
b = np.zeros((3,3))
c = np.ones((4,4))
l = []
for i in range(19):
  if (i%2 == 0):
    l.append(i)

print(f"an array: {a}")
print(f"a 3D array filled with zeros: {b}")
print(f"a 4D array filled with ones: {c}")
# print(l)
d = np.array(l)
print(f"an array of even numbers: {d}")
e = np.linspace(0,1,7)
print(f"numbers evenly spaced between 0 and 1: {e}")
```

flatten() → returns a copy
ravel() → returns a view (faster, uses less memory)

```python
a = np.array([[1, 2, 3], [4, 5, 6]])
flat = a.flatten()
print(flat)
flat = a.ravel()
print(flat)
flat = a.reshape(-1)
print(flat)
```

dot and @ for matric multiplication

```python
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])
C = A * B
print(C)
D = np.dot(A, B)
print(D)
E = A @ B
print(E)
```

```python
grades = np.array([12,15,27,28,13,19])
print(f"mean {np.mean(grades)}")
print(f"median {np.median(grades)}")
print(f"std {np.std(grades)}")
print(f"sum {np.sum(grades)}")
```

Broadcasting

A way numpy automatically expands arrays of diff shape to make operations possible

small array is streched to match with the larger one


```python
a = np.array([1,2,3])
b = np.array([[10],[20],[30]])
c = a+b
print(c)
```

```python
a = np.array([1,2,3])
b = np.array([[10,20,30],[40,50,60],[70,80,90]])
c = a+b
print(c)
```

Line plot

```python
import matplotlib.pyplot as plt
grades = [65,70,80,45,89]
semesters = [1,2,3,4,5]

plt.plot(semesters,grades,marker="s")
plt.title("Student grade over semesters")
plt.xlabel("semester")
plt.ylabel("grades")
plt.show()
```

Bar chart

```python
groups = ["A","B","C","D","F"]
students = [5,10,12,7,3]
plt.bar(groups,students,color="green")
plt.title("Number of students in each grade group")
plt.xlabel("Grade group")
plt.ylabel("Number of students")
plt.show()
```

Pie chart

```python
def dec():
  return "%1.1f%%"

groups = ["A","B","C","D","F"]
students = [5,10,12,7,3]
plt.pie(students,labels=groups,autopct=dec(),startangle=90)
plt.title("Grade distribution")
plt.show()
```

legend and gridlines

```python
x = [1,2,3,4,5]
math_scores = [65,70,75,80,85]
english_scores = [60,68,72,78,82]

plt.plot(x,math_scores,label="Math",color="blue")
plt.plot(x,english_scores,label="English",color="red")
plt.title("NStudent scores over time")
plt.xlabel("Test number")
plt.ylabel("Score")
plt.legend()
plt.grid(True)
plt.show()
```

Subplots

multiple charts inside one window
plt.subplot(rows,cols,index

```python
x = [1,2,3,4,5]
math_scores = [65,70,75,80,85]
english_scores = [60,68,72,78,82]
science_scores = [65,58,82,98,22]

plt.subplot(1,3,1)
plt.plot(x,math_scores,color="blue")
plt.title("Math")

plt.subplot(1,3,2)
plt.plot(x,english_scores,color="red")
plt.title("English")

plt.subplot(1,3,3)
plt.plot(x,science_scores,color="green")
plt.title("Science")

plt.tight_layout() # avoid overlap
plt.show()
```
